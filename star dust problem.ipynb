```python
def fibonacci(n):
    fib_series = [7, 8]  # Initialize the Fibonacci series with the first two terms
    while len(fib_series) < n:
        next_term = fib_series[-1] + fib_series[-2]  # Calculate the next term
        fib_series.append(next_term)  # Add the next term to the series
    return fib_series

# Example usage:
num_terms = 4  # Change this to generate Fibonacci series with a different number of terms
print("Fibonacci series with", num_terms, "terms:")
print(fibonacci(num_terms))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def gaussian_beam(x, mu, sigma):
    return 1/(np.sqrt(2*np.pi)*sigma) * np.exp(-(x - mu)**2 / (2*sigma**2))

mu = 0  
sigma = 1  

x = np.linspace(-5, 5, 100)
y = gaussian_beam(x, mu, sigma)


plt.figure(figsize=(9, 3))
plt.plot(x, y, label='Gaussian Beam Function', color='black')
plt.title('Gaussian Beam Function')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
wavelength = 550e-9  # Wavelength of light (in meters)
a = 5e-6  # Width of the slit (in meters)
I0 = 1  # Maximum intensity

# Define function for intensity
def intensity(theta):
    alpha = np.pi * a * np.sin(theta) / wavelength
    return I0 * (np.sin(alpha) / alpha)**2

# Generate theta values
theta = np.linspace(-np.pi/2, np.pi/2, 10000)

# Calculate intensity values
intensity_values = intensity(theta)

# Plot intensity graph
plt.figure(figsize=(6, 6))
plt.plot(theta, intensity_values, color='blue')
plt.title("Intensity Pattern for Young's Single Slit diffration")
plt.xlabel("Theta (radians)")
plt.ylabel("Intensity")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Constants
wavelength = 550e-9  # Wavelength of light (in meters)
a = 5e-6  # Width of each slit (in meters)
d = 5e-6  # Distance between the slits (in meters)
I0 = 1  # Maximum intensity

# Define function for intensity
def intensity(theta):
    alpha = np.pi * a * np.sin(theta) / wavelength
    beta = np.pi * d * np.sin(theta) / wavelength
    return I0 * (np.cos(beta))**2 * (np.sin(alpha) / alpha)**2

# Generate theta values
theta = np.linspace(-np.pi/2, np.pi/2, 700)

# Calculate intensity values
intensity_values = intensity(theta)

# Plot intensity graph
plt.figure(figsize=(8, 6))
plt.plot(theta, intensity_values, color='blue')
plt.title("Intensity Pattern for Double-Slit Interference")
plt.xlabel("Theta (radians)")
plt.ylabel("Intensity")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
sigma = 1.0  # Standard deviation of the Gaussian beam
x = np.linspace(-5, 5, 1000)  # Range for x-axis

# Gaussian-shaped TE mode intensity profile
electric_field = np.exp(-x**2 / (2 * sigma**2))

# Plot the TE mode intensity profile
plt.figure(figsize=(8, 6))
plt.plot(x, electric_field, color='blue')
plt.title('Transverse Electric Mode Intensity Profile')
plt.xlabel('Position')
plt.ylabel('Electric Field Intensity')
plt.grid(True)
plt.show()
```
