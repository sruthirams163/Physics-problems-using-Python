```python
import numpy as np
import matplotlib.pyplot as plt

def random_walk(n_steps, n_walkers):
    steps = 2 * np.random.randint(2, size=(n_walkers, n_steps)) - 1
    positions = np.cumsum(steps, axis=1)
    squared_deviation = positions**2
    mean_squared_deviation = np.mean(squared_deviation, axis=0)
    return mean_squared_deviation

n_steps_list = [8, 16, 32, 64, 96, 100, 200, 300, 400, 512, 625]

n_walkers = 1000

msd_list = []


for n_steps in n_steps_list:
    msd = random_walk(n_steps, n_walkers)
    msd_list.append(msd)

plt.figure(figsize=(10, 6))
for i, n_steps in enumerate(n_steps_list):
    plt.plot(range(1, n_steps + 1), msd_list[i], label=f'{n_steps} steps')

plt.xlabel('Number of Steps')
plt.ylabel('Mean Square Deviation')
plt.title('Mean Square Deviation vs Number of Steps')
plt.legend()
plt.xscale('log')
plt.yscale('log')
plt.grid(True)
plt.show()
```
